```python
import pickle
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, recall_score, roc_auc_score
import warnings
import sys
import os
import contextlib

warnings.filterwarnings("ignore")

@contextlib.contextmanager
def suppress_stderr():
    with open(os.devnull, "w") as devnull:
        old = sys.stderr
        sys.stderr = devnull
        try:
            yield
        finally:
            sys.stderr = old

# Load model + featurizer
with suppress_stderr():
    rf_model = pickle.load(open("model.pkl", "rb"))
    featurizer = pickle.load(open("featurizer.pkl", "rb"))

# Load dataset
df = pd.read_csv("HIV.csv")
smiles = df["smiles"].values
labels = df["HIV_active"].values

# --------------------------
# FIX: Robust Featurization
# --------------------------
feature_size = 1024

def robust_featurize_list(smiles_list):
    X = []
    for s in smiles_list:
        try:
            feat = featurizer.featurize([s])
            if feat.size > 0:
                X.append(feat[0])
            else:
                X.append(np.zeros(feature_size, dtype=int))
        except:
            X.append(np.zeros(feature_size, dtype=int))
    return np.array(X)

with suppress_stderr():
    X = robust_featurize_list(smiles)

y = labels

# Predictions
y_pred = rf_model.predict(X)
y_prob = rf_model.predict_proba(X)[:, 1]

# Metrics
acc = accuracy_score(y, y_pred)
recall = recall_score(y, y_pred)
auc = roc_auc_score(y, y_prob)

acc, recall, auc
```
